### Student Class


```python
class Student:
    def __init__(self, student_id, name, age, marks):
        self.student_id = student_id
        self.name = name
        self.age = age
        self.marks = marks

    def display(self):
        print("Student ID :", self.student_id)
        print("Name       :", self.name)
        print("Age        :", self.age)
        print("Marks      :", self.marks)
```

### Test Class

```python
s1 = Student(101, "Rahul", 20, 85)
s1.display()
```

### Student Manager Class

```python
class StudentManager:

    def __init__(self):
        self.students = []

    # Add Student
    def add_student(self, student):
        self.students.append(student)
        print("Student added successfully.")

    # Display Students
    def display_students(self):
        if len(self.students) == 0:
            print("No students found.")
        else:
            for student in self.students:
                student.display()
                print("---------------------")

    # Search Student
    def search_student(self, student_id):
        for student in self.students:
            if student.student_id == student_id:
                print("Student Found")
                student.display()
                return
        print("Student not found.")

    # Update Student
    def update_student(self, student_id):
        for student in self.students:
            if student.student_id == student_id:
                student.name = input("Enter new name: ")
                student.age = int(input("Enter new age: "))
                student.marks = int(input("Enter new marks: "))
                print("Student updated successfully.")
                return
        print("Student not found.")

    # Delete Student
    def delete_student(self, student_id):
        for student in self.students:
            if student.student_id == student_id:
                self.students.remove(student)
                print("Student deleted successfully.")
                return
        print("Student not found.")
```

```python
manager = StudentManager()
```

### calculate average marks

```python
class StudentManager:

    def __init__(self):
        self.students = []

    # Add Student
    def add_student(self, student):
        self.students.append(student)
        print("Student added successfully.")

    # Display Students
    def display_students(self):
        if len(self.students) == 0:
            print("No students found.")
        else:
            print("\nStudent Details")
            print("------------------------")
            for student in self.students:
                student.display()
                print("------------------------")

    # Search Student
    def search_student(self, student_id):
        for student in self.students:
            if student.student_id == student_id:
                print("Student Found")
                student.display()
                return
        print("Student not found.")

    # Update Student
    def update_student(self, student_id):
        for student in self.students:
            if student.student_id == student_id:
                student.name = input("Enter new name: ")
                student.age = int(input("Enter new age: "))
                student.marks = int(input("Enter new marks: "))
                print("Student updated successfully.")
                return
        print("Student not found.")

    # Delete Student
    def delete_student(self, student_id):
        for student in self.students:
            if student.student_id == student_id:
                self.students.remove(student)
                print("Student deleted successfully.")
                return
        print("Student not found.")

    # Calculate Average Marks
    def calculate_average_marks(self):
        if len(self.students) == 0:
            print("No students found.")
            return

        total_marks = 0

        for student in self.students:
            total_marks += student.marks

        average = total_marks / len(self.students)

        print("Average Marks =", average)
```

```python
manager = StudentManager()
```

```python
manager.add_student(Student(101, "Rahul", 20, 80))
manager.add_student(Student(102, "Anu", 21, 90))
manager.add_student(Student(103, "Arun", 22, 70))
```

```python
manager.calculate_average_marks()
```

### save records

```python
from openpyxl import Workbook

class StudentManager:

    def __init__(self):
        self.students = []

    # Add Student
    def add_student(self, student):
        self.students.append(student)
        print("Student added successfully.")

    # Display Students
    def display_students(self):
        if len(self.students) == 0:
            print("No students found.")
        else:
            print("\n----- Student Details -----")
            for student in self.students:
                student.display()
                print("--------------------------")

    # Search Student
    def search_student(self, student_id):
        for student in self.students:
            if student.student_id == student_id:
                print("Student Found")
                student.display()
                return
        print("Student not found.")

    # Update Student
    def update_student(self, student_id):
        for student in self.students:
            if student.student_id == student_id:
                student.name = input("Enter new name: ")
                student.age = int(input("Enter new age: "))
                student.marks = int(input("Enter new marks: "))
                print("Student updated successfully.")
                return
        print("Student not found.")

    # Delete Student
    def delete_student(self, student_id):
        for student in self.students:
            if student.student_id == student_id:
                self.students.remove(student)
                print("Student deleted successfully.")
                return
        print("Student not found.")

    # Calculate Average Marks
    def calculate_average_marks(self):
        if len(self.students) == 0:
            print("No students found.")
            return

        total_marks = 0

        for student in self.students:
            total_marks += student.marks

        average = total_marks / len(self.students)

        print("Average Marks =", average)

    # Save Records to Excel
    def save_records(self):
        workbook = Workbook()
        sheet = workbook.active
        sheet.title = "Student Records"

        # Header
        sheet.append(["Student ID", "Name", "Age", "Marks"])

        # Student Records
        for student in self.students:
            sheet.append([
                student.student_id,
                student.name,
                student.age,
                student.marks
            ])

        workbook.save("Student_Records.xlsx")
        print("Records saved successfully in Student_Records.xlsx")
```

```python
manager = StudentManager()
```

```python
manager.save_records()
```
